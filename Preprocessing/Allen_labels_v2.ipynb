```python
import pandas as pd

SRC = 'cell_metadata_with_cluster_annotation_WMB10x_20241115.csv'
OUT = 'Allen_institute_labels_v2_classsubclasssupertype.csv'
```

```python
df = pd.read_csv(SRC, low_memory=False)
df.shape
```

```python
keep = ['cell_label', 'cell_barcode', 'barcoded_cell_sample_label', 'library_label',
        'feature_matrix_label', 'entity', 'brain_section_label', 'library_method',
        'region_of_interest_acronym', 'donor_label', 'donor_genotype', 'donor_sex',
        'dataset_label', 'x', 'y', 'cluster_alias', 'subclass', 'supertype', 'class']

df = df[keep].rename(columns={'subclass': 'subclass_id_label',
                              'supertype': 'supertype_id_label',
                              'class': 'class_id_label'})
df.head()
```

```python
df.to_csv(OUT)
```
