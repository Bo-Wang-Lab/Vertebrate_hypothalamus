```python
import numpy as np
import matplotlib.pyplot as plt
import scanpy as sc
from sklearn.decomposition import TruncatedSVD
import pandas as pd
import os
import anndata as ad
import seaborn as sns
from scipy import sparse
```

```python
#input the h5ads you want to combine
dat_list = ['preprocessed_DanioRerio2.h5ad',
            'preprocessed_DanioRerio3.h5ad',
            'preprocessed_DanioRerio4.h5ad',
            'preprocessed_DanioRerio5.h5ad',
            'preprocessed_DanioRerio6.h5ad',
            'preprocessed_DanioRerio7.h5ad',
            'preprocessed_DanioRerio8.h5ad',
            'preprocessed_DanioRerio9.h5ad',
            'preprocessed_DanioRerio10.h5ad',
            'preprocessed_DanioRerio11.h5ad',
            'preprocessed_DanioRerio12.h5ad',
            'preprocessed_DanioRerio13.h5ad',
            'preprocessed_DanioRerio14.h5ad',
            'preprocessed_DanioRerio15.h5ad',
            'preprocessed_DanioRerio16.h5ad',
           ]
tot_dat = sc.read_h5ad('preprocessed_DanioRerio1.h5ad')
tot_dat.X = sparse.csr_matrix(tot_dat.X)
#Manually set name
tot_dat.obs['key'] = 'DarioRerio1'
for item in dat_list:
    dat = sc.read_h5ad(item)
    dat.X = sparse.csr_matrix(dat.X)
    print(len(tot_dat.var_names))
    #manually set this so that the key name is what you want it to be
    dat.obs['key'] = item.split('_')[1].split('.')[0]
    print(item.split('_')[1].split('.')[0])
    tot_dat = ad.concat([tot_dat,dat], join = 'outer')
```
